## Преобразование выражения

Квантор всеобщности можно заменить квантором существования через отрицание:

$$
\forall y\,P(y) \equiv \neg\exists y\,\neg P(y).
$$

Для импликации справедливо:

$$
A(y) \to B(y) \equiv \neg A(y) \lor B(y).
$$

Отрицание импликации имеет вид:

$$
\neg\bigl(A(y) \to B(y)\bigr) \equiv A(y) \land \neg B(y).
$$

Поэтому:

$$
\forall y\bigl(A(y) \to B(y)\bigr)
\equiv
\neg\exists y\bigl(A(y) \land \neg B(y)\bigr).
$$

Важно: равенство

$$
A(y) \to B(y) \equiv A(y) \land \neg B(y)
$$

неверно. Правая часть является отрицанием импликации.

Исходное выражение преобразуется так:

$$
\exists y\bigl(B(y) \land \neg A(y)\bigr)
\land
\neg\exists y\bigl(A(y) \land \neg B(y)\bigr).
$$

Следовательно, оно истинно, если:

$$
A \subseteq B
\quad\text{и}\quad
B \setminus A \ne \varnothing.
$$

Иными словами:

$$
\boxed{A \subsetneq B}
$$
